```python
import pandas as pd
#Assign data
data={'Name':['Sai','Yuvasri','Poovannan',
'Sabharish','Sangamithra','Rithika','Rubasri'],
'Age':[20,18,19,20,18,17,19],
'Gender':['M','F','M','M','F','F','F'],
'Marks':[90,76,'NaN',74,65,'NaN',71]}
#convert into datafram
df=pd.DataFrame(data)
#display data
print("Original DataFrame:\n",df)
#convert average
c=0
avg_sum=0
for ele in df['Marks']:
    if str(ele).isnumeric():
        c+=1
        avg_sum+=ele

# Calculate average after the loop and check for division by zero
if c > 0:
    avg = avg_sum / c
else:
    avg = 0 # Or handle this case as appropriate, e.g., keep NaN or another default

#Replace missing values and explicitly handle downcasting for future compatibility
df=df.replace(to_replace="NaN",value=avg).infer_objects(copy=False)
#Display data
print("\nReplacing NaN with average marks:\n",df)
```

```python
#Filter top scoring students
df=df[df['Marks']>=75]

#Remove age row
df=df.drop(['Age'],axis=1)

#Display data
print("\nFiltering and Dropping func:\n",df)
```
